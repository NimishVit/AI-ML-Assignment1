```python
import pandas as pd
import numpy as np

df = pd.read_csv('data.csv')

display(df.head())

target_variable = 'diagnosis'
numerical_features = df.select_dtypes(include=[np.number]).columns.tolist()
print(f'Target Variable: {target_variable}')
print(f'Number of Numerical Features: {len(numerical_features)}')

print('Dataset Information')
df.info()

print('Summary Statistics')
display(df.describe())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

print('Missing values in each column:\n', df.isnull().sum())

cols_to_drop = ['id']
if 'Unnamed: 32' in df.columns:
  cols_to_drop.append('Unnamed: 32')

df.drop(columns=cols_to_drop, inplace=True, errors='ignore')

le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])

X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print(f'Training shape: {X_scaled.shape}')
print(f'Testing data shape: {X_test.shape}')
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
print('Predictions generated successfully.')
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f'Accuracy Score: {accuracy:.4f}')
print(f'Precision: {precision:.4f}')
print(f'Recall: {recall:.4f}')
print(f'F1-Score: {f1:.4f}')

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels = ['Benign (0)', 'Malignant (1)'],
            yticklabels=['Benign (0)', 'Malignant (1)'])
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

Conclusion:
In this assignment, we successfully developed a K-Nearest Neighbors (KNN) classifier to predict breast cancer diagnoses. Key findings show that the model achieves excellent performance metrics (Accuracy, Precision, Recall, and F1-Score) when identifying whether a tumor is Malignant or Benign based on diagnostic measurements.

Feature scaling (standardization) played a critical role in this methodology. Because KNN relies on calculating Euclidean distances between data points, features with larger numeric ranges would disproportionately dominate the distance calculations if left unscaled.

However, one notable limitation of the KNN algorithm is its computational expense during the testing phase. Because it does not explicitly build a generalizable internal model, it must store all training data and compute the distance to every single data point for each new prediction, making it slow and memory-intensive for massive datasets.
